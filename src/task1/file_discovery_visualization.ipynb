# Task 1: File Discovery Visualization
Notebook này sẽ đếm số lượng file Python trong thư mục `diffusers` và vẽ biểu đồ phân bố.

```python
import os
import matplotlib.pyplot as plt
import pandas as pd

# Đảm bảo thư mục output tồn tại
output_dir = os.path.abspath(os.path.join(os.getcwd(), '../../data/task1'))
os.makedirs(output_dir, exist_ok=True)

repo_path = '../../diffusers'
excluded_dirs = ['tests', 'docs', 'utils']
excluded_files = ['setup.py']

dir_counts = {}

for root, dirs, files in os.walk(repo_path):
    dirs[:] = [d for d in dirs if d not in excluded_dirs and not d.startswith('.')]
    count = 0
    for file in files:
        if file.endswith('.py') and file not in excluded_files:
            count += 1
    if count > 0:
        # Lấy tên thư mục tương đối
        rel_path = os.path.relpath(root, repo_path)
        dir_counts[rel_path] = count

# Vẽ biểu đồ top 10 thư mục nhiều file python nhất
df = pd.DataFrame(list(dir_counts.items()), columns=['Directory', 'File Count'])
df = df.sort_values(by='File Count', ascending=False).head(10)

plt.figure(figsize=(10, 6))
plt.barh(df['Directory'], df['File Count'], color='skyblue')
plt.xlabel('Số lượng file .py')
plt.ylabel('Thư mục')
plt.title('Top 10 thư mục chứa nhiều file Python nhất trong diffusers')
plt.gca().invert_yaxis()
plt.tight_layout()

# Lưu ảnh
output_file = os.path.join(output_dir, 'file_distribution.png')
plt.savefig(output_file, dpi=300)
print(f'Đã lưu biểu đồ vào {output_file}')
plt.show()
```
